```python
import fastf1
import pandas as pd 
import numpy as np 

session_data = fastf1.get_session(2025, 2, 1)
session_data.load()
laps = session_data.laps
weather_df = laps.get_weather_data().dropna().drop_duplicates()
weather_df = weather_df.rename(columns={'Time':'time', 'AirTemp': 'air_temp', 'Humidity': 'humidity', 'Pressure': 'pressure', 'Rainfall':'is_rainfall', 'TrackTemp':'track_temo',
       'WindDirection':'wind_direction', 'WindSpeed':'wind_speed'})
weather_df
```

```python
weather_records = weather_df.to_dict('records')
weather_records
```

```python
import fastf1.api
df = pd.DataFrame(fastf1.api.race_control_messages(session_data.api_path))
df = df[['Time', 'Message']].rename(columns={'Time':'time', 'Message':'message'}).copy()
df['date'] = df['time']
```
